## **Basis of Numpy Library**

Numpy is the core library for scientific computing in Python. It provides a high-performance multidimensional array object and tools for working with these **arrays**.

To use Numpy, we first need to import the `numpy` package:

```python
# import library
import numpy as np
```

### **Arrays**

- A numpy array is a grid of values, all of the same type, and is indexed by a tuple of nonnegative integers.

- The number of dimensions is the rank of the array;

- The shape of an array is a tuple of integers giving the size of the array along each dimension.

```python
# create 1d array
arr_1d = np.array([1, 2, 3])
```

```python
# print shape
print(arr_1d.shape)
```

```python
# create 2d array
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
# print shape
print(arr_2d.shape)
```

```python
# create 3d array
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
```

```python
# print shape
print(arr_3d.shape)
```

### **Array Indexing**

```python
# print the second element in your 1d array
print(arr_1d[1])
```

```python
# print the first row second element in your 2d array
print(arr_2d[0, 1])
```

```python
# print the first rank first row second element in your 3d array
print(arr_3d[0, 0, 1])
```

### **Array Operations**

```python
# create 2 2d-array
x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])

# Elementwise sum; both produce an array
print(x + y)
```

```python
# Elementwise difference; both produce an array
print(x - y)
```

```python
# Elementwise division; both produce an array
print(x / y)
```

```python
# Elementwise product; both produce the array
print(x * y)
```

```python
# Dot product or matrix multiplication
print(np.dot(x, y))
```

```python
# read file
import cv2
from google.colab.patches import cv2_imshow

filename = '/content/drive/MyDrive/Colab Notebooks/113001234_林小明_HW2.png' # your file path

# 讀取圖片
img = cv2.imread(filename)

# 在 Colab 顯示原圖確認是否讀取成功
cv2_imshow(img)
```

```python
# Rotate
# 讀取圖片的高與寬 (使用 [:2] 可避免 X 光片若為灰階圖而缺少深度維度導致報錯)
(h, w) = img.shape[:2]
center = (w // 2, h // 2) # Find the center of image

# 第一個參數為旋轉軸心，第二個參數為旋轉角度，第三個參數為縮放比例
# 這裡依照簡報範例，順時針旋轉 30 度
M = cv2.getRotationMatrix2D(center, -180, 1.0)

# 將旋轉後的圖片投射在一張布幕上，第三個參數為布幕大小
img_rotated = cv2.warpAffine(img, M, (w, h))

cv2_imshow(img_rotated)
```

```python
# Denoise by filter
# 使用高斯濾波器進行去噪，(5, 5) 為 kernel size，可視去噪效果自行微調
img_denoised = cv2.GaussianBlur(img_rotated, (5, 5), 0)

# 顯示去噪後的最終圖片
cv2_imshow(img_denoised)

# 將最終處理好的圖片存回 Google Drive，方便下載並上傳至 eeclass
# 請務必將檔名替換成您的學號與姓名
save_path = '/content/drive/MyDrive/Colab Notebooks/114066522_李植雋_HW2.png'
cv2.imwrite(save_path, img_denoised)
print("圖片已儲存至:", save_path)
```
